```python
import numpy as np
import cvxpy as cp

# --- Data Setup for the Challenging Problem ---
currencies = ['USD', 'EUR', 'GBP', 'JPY', 'CHF'] 
n = len(currencies)

# 1. Initial Holdings (in native units) - SCARCE & DIVERSE
c_init = np.array([100000.0,   # USD
                   80000.0,    # EUR
                   50000.0,    # GBP
                   1000000.0,  # JPY (REDUCED! Cannot cover requirement alone)
                   100000.0])       # CHF

# 2. Required Holdings (in native units) - ONLY CHF
c_req = np.array([50000.0,        # USD
                  60000.0,        # EUR
                  60000.0,        # GBP
                  600000.0,        # JPY
                  150000.0])  # CHF

# 3. MANIPULATED Exchange Rate Matrix (R[i, j])
R = np.array([[1.00,   0.90,   0.80,   0.009,  0.99],   # USD
              [1.10,   1.00,   0.90,   0.010,  1.08],   # EUR
              [1.25,   1.11,   1.00,   0.011,  1.22],   # GBP
              [111.0,  100.0,  90.0,   1.00,   120.0],  # JPY
              [1.00,   0.91,   0.80,   0.008,  1.00]])  # CHF

# 4. MANIPULATED Transaction Cost Matrix (Delta[i, j])
# High cost (2%) for any trade involving CHF. Low cost (0.1%) for others.
Delta = np.array([[0.0000, 0.0010, 0.0010, 0.0010, 0.0200],
                  [0.0010, 0.0000, 0.0010, 0.0010, 0.0200],
                  [0.0010, 0.0010, 0.0000, 0.0010, 0.0200],
                  [0.0010, 0.0010, 0.0010, 0.0000, 0.0200],
                  [0.0200, 0.0200, 0.0200, 0.0200, 0.0000]])

# --- Define and Solve the Optimization Problem ---
X = cp.Variable((n, n), nonneg=True)
constraints = [cp.diag(X) == 0] # No self-trades

# Net flow constraint: Final amount >= Required amount
inflow = cp.sum(cp.multiply(X, R), axis=0)
outflow = cp.sum(X, axis=1)
final_holdings = c_init + inflow - outflow
constraints.append(final_holdings >= c_req)

# NEW: Calculate total cost in a single base currency (USD).
# How much USD is 1 unit of currency j worth? We can use the inverse of R[0, j] (USD per j).
# Example: 1 EUR is worth 1/0.85 ≈ 1.176 USD (according to our R matrix)
usd_per_unit = np.array([1.0, 1/0.85, 1/0.75, 1/0.0068, 1/1.02])

# The cost of a trade X[i,j] is X[i,j] * Δ[i,j] units of currency j.
# The USD value of that cost is (X[i,j] * Δ[i,j]) * (USD value of 1 unit of j)
# We need to multiply each COLUMN j of the (X * Delta) matrix by usd_per_unit[j]
# We can do this by creating a 2D array where each row is the usd_per_unit vector,
# and then multiply element-wise with the (X * Delta) matrix.
usd_per_unit_matrix = np.tile(usd_per_unit, (n, 1))  # Repeat the vector to form a n x n matrix
total_cost_usd = cp.sum(cp.multiply(cp.multiply(X, Delta), usd_per_unit_matrix))

# Let's minimize the cost in USD, which is meaningful.
objective = cp.Minimize(total_cost_usd)
prob = cp.Problem(objective, constraints)
result = prob.solve(verbose=False)

# --- Print the Results ---
print("\n" + "="*50)
print("OPTIMIZATION: CROSS-BORDER PAYMENT")
print("="*50)
print(f"Status: {prob.status}")
print(f"Total Transaction Cost: ${total_cost_usd.value:,.2f} USD")

print("\n" + "-"*50)
print("FINAL HOLDINGS")
print("-"*50)
print(f"{'Currency':<6} {'Initial':>12} {'Required':>12} {'Final':>12}")
for i, curr in enumerate(currencies):
    print(f"{curr:<6} {c_init[i]:>12,.2f} {c_req[i]:>12,.2f} {final_holdings[i].value:>12,.2f}")

print("\n" + "-"*50)
print("OPTIMAL TRADING PLAN (Amount of j Sold)")
print("-"*50)
header = " Receive \\ Sell |"
for j in range(n):
    header += f" {currencies[j]:>10}"
print(header)
print("-" * (16 + 11*n))
for i in range(n):
    row_str = f"{currencies[i]:<14} |"
    for j in range(n):
        value = 0.0 if i == j else X.value[i, j]
        row_str += f" {value:>10.2f}"
    print(row_str)
```

```python
import numpy as np
import cvxpy as cp

# --- Data Setup ---
currencies = ['USD', 'EUR', 'GBP', 'JPY', 'CHF'] 
n = len(currencies)

# 1. Initial Holdings
c_init = np.array([100000.0,   # USD
                   80000.0,    # EUR
                   50000.0,    # GBP
                   1000000.0,  # JPY
                   100000.0])  # CHF

# 2. Required Holdings
c_req = np.array([50000.0,     # USD
                  60000.0,     # EUR
                  60000.0,     # GBP
                  600000.0,    # JPY
                  150000.0])   # CHF

# 3. NOMINAL Exchange Rate Matrix
R_nom = np.array([[1.00,   0.90,   0.80,   0.009,  0.99],
                  [1.10,   1.00,   0.90,   0.010,  1.08],
                  [1.25,   1.11,   1.00,   0.011,  1.22],
                  [111.0,  100.0,  90.0,   1.00,   120.0],
                  [1.00,   0.91,   0.80,   0.008,  1.00]])

# 4. NOMINAL Transaction Cost Matrix
Delta_nom = np.array([[0.0000, 0.0010, 0.0010, 0.0010, 0.0200],
                      [0.0010, 0.0000, 0.0010, 0.0010, 0.0200],
                      [0.0010, 0.0010, 0.0000, 0.0010, 0.0200],
                      [0.0010, 0.0010, 0.0010, 0.0000, 0.0200],
                      [0.0200, 0.0200, 0.0200, 0.0200, 0.0000]])

# --- Define Uncertainty Radii ---
# Assume 2% uncertainty for exchange rates
rho_R = 0.02 * R_nom
# Avoid negative radii for zero elements (like diagonals)
rho_R = np.maximum(rho_R, 0.0)

# Assume 10% uncertainty for transaction costs
rho_Delta = 0.10 * Delta_nom
rho_Delta = np.maximum(rho_Delta, 0.0)

print("Exchange Rate Uncertainty Radii (rho_R):")
print(rho_R)
print("\nTransaction Cost Uncertainty Radii (rho_Delta):")
print(rho_Delta)
```

```python
# --- Formulate the Robust Counterpart ---
X = cp.Variable((n, n), nonneg=True)
constraints = [cp.diag(X) == 0]  # No self-trades

# 1. Use WORST-CASE exchange rates for constraints (lower bound)
R_worst = R_nom - rho_R
inflow = cp.sum(cp.multiply(X, R_worst), axis=0)
outflow = cp.sum(X, axis=1)
final_holdings = c_init + inflow - outflow
constraints.append(final_holdings >= c_req)

# 2. Use WORST-CASE transaction costs for objective (upper bound)
Delta_worst = Delta_nom + rho_Delta
# Calculate total cost in USD (using nominal R for cost conversion, or worst-case? Here we use nominal)
usd_per_unit = np.array([1.0, 1/0.90, 1/0.80, 1/0.009, 1/0.99])  # Based on first row of R_nom
usd_per_unit_matrix = np.tile(usd_per_unit, (n, 1))
total_cost_usd = cp.sum(cp.multiply(cp.multiply(X, Delta_worst), usd_per_unit_matrix))

objective = cp.Minimize(total_cost_usd)
prob = cp.Problem(objective, constraints)

# --- Solve the Robust Problem ---
result = prob.solve(verbose=False)

# --- Print Results ---
print("\n" + "="*60)
print("ROBUST OPTIMIZATION RESULTS (Box Uncertainty)")
print("="*60)
print(f"Solver Status: {prob.status}")
print(f"Worst-case Total Transaction Cost: ${total_cost_usd.value:,.2f} USD")

# Calculate final holdings achieved under the worst-case rates
final_holdings_achieved = c_init + np.sum(X.value * R_worst, axis=0) - np.sum(X.value, axis=1)
print("\nFINAL HOLDINGS (Under Worst-case Exchange Rates):")
print(f"{'Currency':<6} {'Initial':>10} {'Required':>10} {'Final':>12} {'Surplus/Deficit':>15}")
for i, curr in enumerate(currencies):
    surplus_def = final_holdings_achieved[i] - c_req[i]
    print(f"{curr:<6} {c_init[i]:>10,.0f} {c_req[i]:>10,.0f} {final_holdings_achieved[i]:>12,.2f} {surplus_def:>15,.2f}")

print("\nROBUST TRADING PLAN (Amount of j Sold):")
header = " Buy \\ Sell |"
for j in range(n):
    header += f" {currencies[j]:>10}"
print(header)
print("-" * (12 + 11*n))
for i in range(n):
    row_str = f"{currencies[i]:<10} |"
    for j in range(n):
        value = 0.0 if i == j else X.value[i, j]
        row_str += f" {value:>10.2f}"
    print(row_str)

# For comparison, let's calculate the nominal cost of this robust plan
nominal_cost_usd = cp.sum(cp.multiply(cp.multiply(X, Delta_nom), usd_per_unit_matrix)).value
print(f"\nNominal Cost of Robust Plan: ${nominal_cost_usd:,.2f} USD")
```
