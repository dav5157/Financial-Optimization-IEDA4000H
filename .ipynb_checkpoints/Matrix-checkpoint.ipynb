```python
def matrix_multiply(A, B):
    # Check if matrices can be multiplied
    if len(A[0]) != len(B):
        return "Error: Number of columns in first matrix must equal number of rows in second matrix"
    
    # Initialize result matrix with zeros
    result = [[0 for _ in range(len(B[0]))] for _ in range(len(A))]
    
    # Perform matrix multiplication
    for i in range(len(A)):
        for j in range(len(B[0])):
            for k in range(len(B)):
                result[i][j] += A[i][k] * B[k][j]
    
    return result

def print_matrix(matrix):
    # Print matrix in a readable format
    for row in matrix:
        print([f"{x:4}" for x in row])

def main():
    print("Matrix Multiplication Calculator")
    print("\nInput Syntax:")
    print("- Enter matrices as comma-separated numbers with rows separated by semicolons")
    print("- Example: 1,2,3;4,5,6 for a 2x3 matrix [[1,2,3],[4,5,6]]")
    print("- Matrices must be compatible for multiplication (columns of A = rows of B)")
    
    try:
        # Get input for first matrix
        matrix1_input = input("\nEnter first matrix: ")
        matrix1 = [[int(x) for x in row.split(',')] for row in matrix1_input.split(';')]
        
        # Get input for second matrix
        matrix2_input = input("Enter second matrix: ")
        matrix2 = [[int(x) for x in row.split(',')] for row in matrix2_input.split(';')]
        
        print("\nMatrix A:")
        print_matrix(matrix1)
        print("\nMatrix B:")
        print_matrix(matrix2)
        
        # Perform multiplication
        result = matrix_multiply(matrix1, matrix2)
        
        if isinstance(result, str):
            print(f"\n{result}")
        else:
            print("\nResult (A × B):")
            print_matrix(result)
            
    except ValueError:
        print("Error: Invalid input format. Please use numbers separated by commas and rows by semicolons")
    except Exception as e:
        print(f"Error: {str(e)}")

if __name__ == "__main__":
    main()
```
